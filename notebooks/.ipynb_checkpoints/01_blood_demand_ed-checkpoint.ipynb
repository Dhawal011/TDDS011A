```python
df_clean["SUPPLY_GAP"] = (
    df_clean["QTY_SUPPLIED"] -
    df_clean["QTY_DEMANDED"]
)
```

```python
df_clean["SHORTAGE"] = (
    df_clean["SUPPLY_GAP"] < 0
).astype(int)
```

```python
print(df_clean["SHORTAGE"].value_counts())
```

```python
print(
    df_clean[
        df_clean["SHORTAGE"] == 1
    ][
        ["DATE", "QTY_DEMANDED", "QTY_SUPPLIED", "SUPPLY_GAP"]
    ]
)
```

```python
ml_df = df_clean.copy()

ml_df = ml_df.sort_values("DATE").reset_index(drop=True)

ml_df["LAG_1"] = ml_df["QTY_DEMANDED"].shift(1)
ml_df["LAG_2"] = ml_df["QTY_DEMANDED"].shift(2)
ml_df["LAG_3"] = ml_df["QTY_DEMANDED"].shift(3)

ml_df["ROLLING_MEAN_3"] = (
    ml_df["QTY_DEMANDED"]
    .shift(1)
    .rolling(window=3)
    .mean()
)

ml_df["ROLLING_MEAN_6"] = (
    ml_df["QTY_DEMANDED"]
    .shift(1)
    .rolling(window=6)
    .mean()
)

ml_df.head(10)
```

```python
ml_df["MONTH_NUMBER"] = ml_df["DATE"].dt.month
ml_df["YEAR_NUMBER"] = ml_df["DATE"].dt.year

ml_df.head()
```

```python
ml_df["MONTH_SIN"] = np.sin(
    2 * np.pi * ml_df["MONTH_NUMBER"] / 12
)

ml_df["MONTH_COS"] = np.cos(
    2 * np.pi * ml_df["MONTH_NUMBER"] / 12
)
```

```python
ml_df = ml_df.dropna().reset_index(drop=True)
```

```python
print("Original rows:", len(df_clean))
print("ML rows:", len(ml_df))

print("\nMissing values:")
print(ml_df.isna().sum())
```

```python
ml_df[
    [
        "DATE",
        "QTY_DEMANDED",
        "LAG_1",
        "LAG_2",
        "LAG_3",
        "ROLLING_MEAN_3",
        "ROLLING_MEAN_6",
        "MONTH_NUMBER",
        "MONTH_SIN",
        "MONTH_COS"
    ]
].head(15)
```

```python
train = ml_df[
    ml_df["DATE"] < "2019-01-01"
].copy()

test = ml_df[
    ml_df["DATE"] >= "2019-01-01"
].copy()
```

```python
print("Training period:")
print(train["DATE"].min(), "to", train["DATE"].max())

print("\nTesting period:")
print(test["DATE"].min(), "to", test["DATE"].max())

print("\nTraining samples:", len(train))
print("Testing samples:", len(test))
```

```python
features = [
    "LAG_1",
    "LAG_2",
    "LAG_3",
    "ROLLING_MEAN_3",
    "ROLLING_MEAN_6",
    "MONTH_NUMBER",
    "MONTH_SIN",
    "MONTH_COS"
]
```

```python
X_train = train[features]
y_train = train["QTY_DEMANDED"]

X_test = test[features]
y_test = test["QTY_DEMANDED"]

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
linear_model = LinearRegression()

linear_model.fit(
    X_train,
    y_train
)
```

```python
linear_predictions = linear_model.predict(X_test)
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)

linear_mae = mean_absolute_error(
    y_test,
    linear_predictions
)

linear_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        linear_predictions
    )
)

print("Linear Regression")
print("-----------------")
print("MAE :", round(linear_mae, 2))
print("RMSE:", round(linear_rmse, 2))
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    test["DATE"],
    y_test,
    marker="o",
    label="Actual"
)

plt.plot(
    test["DATE"],
    linear_predictions,
    marker="o",
    label="Predicted"
)

plt.title("Linear Regression — Actual vs Predicted Blood Demand")
plt.xlabel("Date")
plt.ylabel("Blood Demand")
plt.legend()
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf_model = RandomForestRegressor(
    n_estimators=300,
    max_depth=6,
    min_samples_leaf=2,
    random_state=42
)
```

```python
rf_model.fit(
    X_train,
    y_train
)
```

```python
rf_predictions = rf_model.predict(X_test)
```

```python
rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_predictions
    )
)

print("Random Forest")
print("-------------")
print("MAE :", round(rf_mae, 2))
print("RMSE:", round(rf_rmse, 2))
```

```python
model_results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest"
    ],
    "MAE": [
        linear_mae,
        rf_mae
    ],
    "RMSE": [
        linear_rmse,
        rf_rmse
    ]
})

model_results.round(2)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    test["DATE"],
    y_test,
    marker="o",
    label="Actual"
)

plt.plot(
    test["DATE"],
    rf_predictions,
    marker="o",
    label="Random Forest"
)

plt.title(
    "Random Forest — Actual vs Predicted Blood Demand"
)

plt.xlabel("Date")
plt.ylabel("Blood Demand")
plt.legend()

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
model_results = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest",
        "Gradient Boosting"
    ],
    "MAE": [
        linear_mae,
        rf_mae,
        gb_mae
    ],
    "RMSE": [
        linear_rmse,
        rf_rmse,
        gb_rmse
    ]
})

model_results.round(2)
```

```python
from sklearn.metrics import mean_absolute_percentage_error
```

```python
linear_mape = mean_absolute_percentage_error(
    y_test,
    linear_predictions
) * 100

rf_mape = mean_absolute_percentage_error(
    y_test,
    rf_predictions
) * 100

gb_mape = mean_absolute_percentage_error(
    y_test,
    gb_predictions
) * 100

model_results["MAPE"] = [
    linear_mape,
    rf_mape,
    gb_mape
]

model_results.round(2)
```
