```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", None)
```

```python
df = pd.read_csv("../data/raw/Blood_data_Tema.csv")

df.head()
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df["DATE"] = pd.to_datetime(
    df["YEAR"].astype(str) + "-" + df["MONTH"],
    format="%Y-%B"
)

df = df.sort_values("DATE").reset_index(drop=True)

df.head()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df["DATE"],
    df["QTY_DEMANDED"],
    marker="o"
)

plt.title("Monthly Blood Demand")
plt.xlabel("Date")
plt.ylabel("Quantity Demanded")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    df["DATE"],
    df["QTY_DEMANDED"],
    marker="o",
    label="Demand"
)

plt.plot(
    df["DATE"],
    df["QTY_SUPPLIED"],
    marker="o",
    label="Supply"
)

plt.title("Monthly Blood Demand vs Supply")
plt.xlabel("Date")
plt.ylabel("Quantity")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df["SUPPLY_GAP"] = df["QTY_SUPPLIED"] - df["QTY_DEMANDED"]

df[[
    "DATE",
    "QTY_DEMANDED",
    "QTY_SUPPLIED",
    "SUPPLY_GAP"
]].head(15)
```

```python
shortage_months = df[df["SUPPLY_GAP"] < 0]

print("Potential shortage months:", len(shortage_months))

shortage_months[[
    "DATE",
    "QTY_DEMANDED",
    "QTY_SUPPLIED",
    "SUPPLY_GAP"
]]
```
